```python
df2=pd.read_csv('/content/students.csv', delimiter=',')
```

```python
df2['Brother-sister'].value_counts()
```

```python
singles=df2[df2['Brother-sister']=='все, у кого нет ни брата ни сестры']
```

```python
sublings=df2[df2['Brother-sister']!='нет ни брата ни сестры']
```

```python
singles['Children number'].median()
```

```python
sublings['Children number'].median()
```

```python
singles['Children number']=singles['Children number'].fillna(0)
```

```python
sublings['Children number']=sublings['Children number'].fillna(2)
```

```python
df_new2=pd.concat([singles, sublings])
df_new2.info()
```

```python
df_new2['Children number'].median()
```
