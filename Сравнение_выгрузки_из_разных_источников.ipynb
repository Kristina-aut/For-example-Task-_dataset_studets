```python
merged_df=pd.merge(
    result_apteka,
    df_t_doc_cut,
    left_on='Номер накладной',
    right_on='Номер',
    how='left'
)
```

```python
merged_df.rename(columns={
    'Номер': 'Номер счета-фактуры',
    'Сумма': 'Сумма счет-фактуры',
    'Дата': 'Дата счет-фактуры'
}, inplace=True)
```

```python
cols = merged_df.columns.tolist()
insert_pos=cols.index('Номер накладной') + 1
```

```python
for col in ['Номер счета-фактуры', 'Сумма счет-фактуры', 'Дата счет-фактуры']:
  cols.remove(col)
```

```python
cols.insert(insert_pos, 'Номер счета-фактуры')
cols.insert(insert_pos + 1, 'Сумма счет-фактуры')
cols.insert(insert_pos + 2, 'Дата счет-фактуры')
```

```python
merged_df=merged_df[cols]
```

```python
merged_df['Дaтa накладной']=pd.to_datetime(merged_df['Дата накладной'], dayfirst=True, errors='coerce').dt.strftime('%d.%m.%Y')
```

```python
merged_df=merged_df.loc[:, ~merged_df.columns.duplicated()]
```

```python
merged_df['Дaтa счет-фактуры']=pd.to_datetime(merged_df['Дата счет-фактуры'], dayfirst=True, errors='coerce').dt.strftime('%d.%m.%Y')
```
