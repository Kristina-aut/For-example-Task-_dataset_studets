```python
a=df['Age'].quantile(0.25)
b=df['Age'].quantile(0.75)
```

```python
df2=df[(df['Age']<b+1.5*(b-a)) & (df['Age']>a-1.5*(b-a))]
sns.boxplot(df['Age'])
```

```python
df['Putin age'].quantile(0.25), df['Putin age'].quantile(0.75)
```

```python
vv=df['Putin age'].quantile(0.25)
dd=df['Putin age'].quantile(0.75)
```

```python
vv-1.5*(dd-vv), dd+1.5*(dd-vv)
```

```python
df3=df[(df['Putin age']<dd+1.5*(dd-vv)) & (df['Putin age']>vv-1.5*(dd-vv))]
sns.boxplot(df['Putin age'])
```
